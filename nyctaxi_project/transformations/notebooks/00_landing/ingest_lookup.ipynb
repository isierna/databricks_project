```python
import urllib.request
import os

from nyctaxi_project.transformations.modules.data_loader.file_downloader import download_file


try:
    url_to_file = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

    response = urllib.request.urlopen(url_to_file)

    if "DATABRICKS_RUNTIME_VERSION" in os.environ:
        # to get the path when run in databricks UI
        dbutils.widgets.text("uploaded_data_path", "/Volumes/nyctaxi/00_landing/data_sources/lookup/")

        dir_path = dbutils.widgets.get("uploaded_data_path")
    else:
        from dotenv import load_dotenv

        load_dotenv() 
        dir_path = os.getenv("LOCAL_UPLOADED_DATA_PATH")
        os.makedirs(dir_path, exist_ok=True)

    destination_path = f"{dir_path}taxi_zone_lookup.csv"

    download_file(url_to_file, destination_path)

    dbutils.jobs.taskValues.set(key="continue_downstream", value="yes")
    print("File succesfully uploaded")
except Exception as e:
    dbutils.jobs.taskValues.set(key="continue_downstream", value="no")
    print(f"File not uploaded {str(e)}")
```
