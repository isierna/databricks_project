```python
from pyspark.sql.functions import date_format
```

```python
# add a year-month column formated as yyyy-MM

df = spark.read.table("nyctaxi.02_silver.yellow_trips_enriched")

df = df.withCoumn("year_month", date_format("tpep_pickup_datetime","yyyy-MM"))
```

```python
# write new dataframe with new coumn in JSON format into EXTERNAL table yellow_trips_export

df.write.\
    option("path", "abfss://nyctaxi-yellow@nyctaxistorage0000001.dfs.core.windows.net/yellow_trips_export/").\
    format("json").\
    mode("overwrite").\
    partitionBy("vendor", "year_month").\
    saveAsTable("nyctaxi.04_export.yellow_trips_export") 
```
