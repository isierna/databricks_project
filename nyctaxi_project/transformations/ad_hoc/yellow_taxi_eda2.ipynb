```python
from pyspark.sql.functions import *
from delta.tables import *

dt = DeltaTable.forName(spark,"nyctaxi.01_bronze.yellow_trips_raw")
dt.delete("tpep_pickup_datetime >= '2025-11-01'")

spark.read.table("nyctaxi.01_bronze.yellow_trips_raw").\
    groupBy(date_format("tpep_pickup_datetime","yyyy-MM").alias("year-month")).\
    agg(count("*").alias("total_records")).\
    orderBy("year-month").\
    display()
```
