```python
from pyspark.sql.functions import sum, col, round

df = spark.read.table("nyctaxi.02_silver.yellow_trips_enriched")
df.display()
```

```python
# time series chart

from pyspark.sql.functions import count, round, sum, col

df.groupBy(
    df.tpep_pickup_datetime.cast("date").alias("date")
).agg(
    round(sum(col("total_amount")), 2).alias("total_amount"), count("*").alias("count")
).sort(col("date").asc()).display()
```

```python
# vendor with highest revenue

df.groupBy(
    df.vendor).\
    agg(round(sum(col("total_amount")), 2).alias("total_amount")).\
    sort(col("total_amount").desc()).limit(1).display()
```

```python
# most popular pickup borough

df.groupBy(
    df.pickup_borough).\
    count().\
    sort(col("count").desc()).limit(1).display()
```

```python
# most common journey

df.groupBy(
    df.pickup_borough, df.dropoff_borough).\
    count().\
    sort(col("count").desc()).limit(1).display()
```
