```python
import urllib.request
import shutil
import os

date_to_process = ['2025-05','2025-06','2025-07','2025-08','2025-09','2025-10','2025-11']

for month in date_to_process:
    url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{month}.parquet"

    response = urllib.request.urlopen(url)

    dir_path = f"/Volumes/nyctaxi/00_landing/data_sources/nyctaxi_yellow/{month}"

    os.makedirs(dir_path, exist_ok=True)

    local_path =dir_path + f"/yellow_tripdata_{month}.parquet"

    with open (local_path, 'wb') as out_file:
        shutil.copyfileobj(response, out_file)
```
