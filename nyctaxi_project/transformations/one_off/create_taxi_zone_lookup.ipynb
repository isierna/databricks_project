```python
import urllib.request
import shutil
import os

url_to_file = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

response = urllib.request.urlopen(url_to_file)

dir_path = "/Volumes/nyctaxi/00_landing/data_sources/lookup"
os.makedirs(dir_path, exist_ok=True)

local_path = "/Volumes/nyctaxi/00_landing/data_sources/lookup/taxi_zone_lookup.csv"

with open (local_path, 'wb') as out_file:
    shutil.copyfileobj(response, out_file)


```
